## 1. Filter your transactions
Keep transactions where amount is greater than 100. Store the result in large.

```python
import pandas as pd

df = pd.DataFrame({
    'category': ['Books', 'Tech', 'Books', 'Tech'],
    'amount': [50, 200, 150, 300]
})

large = df
print(large)
```
