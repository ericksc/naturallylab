## 2. Handle missing values
Remove rows with missing values from df. Store the result in clean and print it.

```python
import pandas as pd

df = pd.DataFrame({
    'name': ['Ana', 'Luis', 'Mia', 'Ana'],
    'score': [90, None, 85, 90]
})

clean = df
print(clean)
```

```python
# Run this cell to check step 2
assert ('clean' in globals() and isinstance(clean, pd.DataFrame) and clean.equals(df.dropna())), "Not quite yet. Review the task and try again."
print("ND_CHECK:clean-1:PASS")
print("Step 2 complete. Continue with the next instruction.")
```

## 3. Remove duplicate rows
Create clean with no missing values and no duplicate rows. Keep the original row indexes.

```python
import pandas as pd

df = pd.DataFrame({
    'name': ['Ana', 'Luis', 'Mia', 'Ana'],
    'score': [90, None, 85, 90]
})

clean = df
print(clean)
```

```python
# Run this cell to check step 3
assert ('clean' in globals() and isinstance(clean, pd.DataFrame) and clean.equals(df.dropna().drop_duplicates())), "Not quite yet. Review the task and try again."
print("ND_CHECK:clean-2:PASS")
print("Step 3 complete. Continue with the next instruction.")
```
